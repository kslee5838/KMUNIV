```python
from urllib.parse import quote_plus

songs = [
    ("Hotel California", "Eagles"),
    ("Imagine", "John Lennon"),
    ("Take Five", "Dave Brubeck")
]

for title, artist in songs:
    query = quote_plus(f"{title} {artist}")
    print(f"{title}: https://open.spotify.com/search/{query}")

```

```python
import pandas as pd

songs = pd.DataFrame({
    "title": [
        "Hotel California",
        "Imagine",
        "Take Five"
    ],
    "artist": [
        "Eagles",
        "John Lennon",
        "Dave Brubeck"
    ]
})
```

```python
search_title = "Imagine"

results = songs[
    songs["title"].str.contains(
        search_title,
        case=False,
        na=False
    )
]
```

```python
from urllib.parse import quote_plus

results = results.copy()

results["spotify_url"] = results.apply(
    lambda row: (
        "https://open.spotify.com/search/"
        + quote_plus(f"{row['title']} {row['artist']}")
    ),
    axis=1
)
```

```python
from IPython.display import display, HTML
from urllib.parse import quote_plus

row = results.iloc[0]

url = (
    "https://open.spotify.com/search/"
    + quote_plus(f"{row['title']} {row['artist']}")
)

display(HTML(
    f'<a href="{url}" target="_blank">'
    f'Open {row["title"]} by {row["artist"]} in Spotify'
    f'</a>'
))
```
